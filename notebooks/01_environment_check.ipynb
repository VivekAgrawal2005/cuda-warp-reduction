```python
import os

PROJECT = "/content/warp_reduction"

folders = [
    "notebooks",
    "src",
    "results",
    "slides"
]

os.makedirs(PROJECT, exist_ok=True)

for folder in folders:
    os.makedirs(os.path.join(PROJECT, folder), exist_ok=True)

print("Project folders created!")
print(PROJECT)
```

```python
import subprocess

def check_command(command):
    result = subprocess.run(
        command,
        shell=True,
        capture_output=True,
        text=True
    )
    return result.stdout.strip() or result.stderr.strip()

gpu_info = check_command("nvidia-smi")
cuda_info = check_command("nvcc --version")

with open(f"{PROJECT}/environment.txt", "w") as f:
    f.write("GPU CHECK\n")
    f.write(gpu_info + "\n\n")
    f.write("NVCC CHECK\n")
    f.write(cuda_info + "\n")

print("Environment report saved.")
```

```python
for root, dirs, files in os.walk(PROJECT):
    level = root.replace(PROJECT, "").count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(root)}/")
    for file in files:
        print(f"{indent}  {file}")
```
